```python
import torch
import torch.nn.functional as F
from pathlib import Path
import random
import matplotlib.pyplot as plt
import torch.nn as nn
import optuna
import pandas
%matplotlib inline
```

```python
path = Path("names.txt")

words = path.read_text().splitlines()

print(len(words))
print(words[:10])
```

```python
chs = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i, s in enumerate(chs)}
stoi['.'] = 0
itos = {i:s for s, i in stoi.items()}
```

```python
def b_ds(words):
    bs = 3
    x, y = [], []
    
    for w in words:
    
        context = [0] * bs
        for ch in w + '.':
            ix = stoi[ch]
            x.append(context)
            y.append(ix)
            context = context[1:] + [ix]
    
    x = torch.tensor(x)
    y = torch.tensor(y)
    return x, y
```

```python
random.shuffle(words)
n1 = int(0.8*len(words))
n2 = int(0.9*len(words))
```

```python
xtrain, ytrain = b_ds(words[:n1])
xval, yval = b_ds(words[n1:n2])
xtest, ytest = b_ds(words[n2:])
```

```python
Xtr, Ytr = xtrain.long().to("cuda"), ytrain.long().to("cuda")
Xva, Yva = xval.long().to("cuda"), yval.long().to("cuda")
```

```python
BLOCK = Xtr.shape[1]
```

```python
class NameMLP(nn.Module):
    def __init__(self, emb_dim, hidden_dim):
        super().__init__()

        self.emb = nn.Embedding(27, emb_dim)
        self.fc1 = nn.Linear(BLOCK * emb_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, 27)

    def forward(self, x):
        x = self.emb(x).flatten(1)
        x = torch.tanh(self.fc1(x))
        return self.fc2(x)
```

```python
@torch.no_grad()
def eval_loss(model, X, Y, chunk=4096):
    was_training = model.training
    model.eval()

    total = 0.0

    for start in range(0, len(Y), chunk):
        end = start + chunk
        logits = model(X[start:end])

        total += F.cross_entropy(
            logits,
            Y[start:end],
            reduction="sum"
        ).item()

    model.train(was_training)
    return total / len(Y)
```

```python
def train_once(cfg, steps=15_000, seed=42,
               trial=None, eval_every=2000):

    torch.manual_seed(seed)

    model = NameMLP(
        cfg["emb_dim"],
        cfg["hidden_dim"]
    ).to("cuda")

    optimizer = torch.optim.SGD(
        model.parameters(),
        lr=cfg["lr"]
    )

    batch_rng = torch.Generator(
        device="cuda"
    ).manual_seed(seed + 10_000)

    best_val = float("inf")
    best_step = 0
    best_weights = None
    history = []

    for step in range(1, steps + 1):

        lr = cfg["lr"] * (
            0.1 if step > 0.7 * steps else 1.0
        )
        optimizer.param_groups[0]["lr"] = lr

        ix = torch.randint(
            len(Xtr),
            (cfg["batch_size"],),
            generator=batch_rng,
            device="cuda"
        )

        logits = model(Xtr[ix])
        loss = F.cross_entropy(logits, Ytr[ix])

        optimizer.zero_grad(set_to_none=True)
        loss.backward()
        optimizer.step()

        if step % eval_every == 0 or step == steps:
            val = eval_loss(model, Xva, Yva)

            history.append((step, loss.item(), val))

            if val < best_val:
                best_val = val
                best_step = step

                best_weights = {
                    k: v.detach().cpu().clone()
                    for k, v in model.state_dict().items()
                }

            if trial is not None:
                trial.report(val, step)

                if trial.should_prune():
                    raise optuna.TrialPruned()

    return best_val, best_step, best_weights, history
```

```python
def objective(trial):
    cfg = {
        "emb_dim": trial.suggest_categorical(
            "emb_dim", [8, 10, 16, 24, 32]
        ),
        "hidden_dim": trial.suggest_categorical(
            "hidden_dim", [100, 200, 300, 500]
        ),
        "batch_size": trial.suggest_categorical(
            "batch_size", [64, 128, 256]
        ),
        "lr": trial.suggest_float(
            "lr", 0.01, 0.3, log=True
        ),
    }

    val, step, _, _ = train_once(
        cfg,
        steps=15_000,
        seed=42,
        trial=trial
    )

    trial.set_user_attr("best_step", step)
    return val
```

```python
study = optuna.create_study(
    study_name="makemore_mlp_v1",
    storage="sqlite:///makemore_mlp.db",
    load_if_exists=True,
    direction="minimize",
    sampler=optuna.samplers.TPESampler(seed=1234),
    pruner=optuna.pruners.MedianPruner(
        n_startup_trials=5,
        n_warmup_steps=4000
    )
)

study.optimize(objective, n_trials=20)

print("Best validation:", study.best_value)
print("Best config:", study.best_params)
```
